```python
# ========================================
# SISTEMA DE PRÉSTAMOS
# Versión 1 - Programación basica y uso de listas
# ========================================

# Lista de alumnos activos en el sistema
alumnos = [
           ["26040200", "Azael", "Ingeniería en Sistemas Computacionales"],
           ["26040201", "Alejandro", "Ingeniería industrial"],
           ["26040202", "Marina", "Ineniería en Ingeniería de Software"],
           ["26040203", "Axel", "Ingeniería en Sistemas Computacionales"],
           ["26040204", "Sofía","Ingeniería en Ingeniería de Software"],
          ]
prestamos=[]
import datetime
necesidades=[]

# Cada valor en la lista de computadoras tiene su lista correspondiente con el
# aula donde esta ubicada, el software instalado y el estado de esa computadora
computadoras =[
               ["PC-LC1-01", "LC1", "Python, VS Code", "Disponible",],
               ["PC-LC1-02", "LC1", "Python, Java, VS Code", "Prestada",],
               ["PC-LC1-03", "LC1", "AutoCAD, MATLAB", "Reservada",],
               ["PC-LC2-01", "LC2", "Python, VS Code", "Disponible",],
               ["PC-LC2-02", "LC2", "Python, Java, VS Code",  "Prestada",],
               ["PC-LC2-03", "LC2", "AutoCAD, MATLAB", "Disponible",],
               ["PC-LC3-01", "LC3", "Python, VS Code", "Disponible",],
               ["PC-LC3-02", "LC3", "Python, Java, VS Code", "Prestada",],
               ["PC-LC3-03", "LC3", "AutoCAD, MATLAB", "Reservada",],
               ["PC-LC3-04"  "LC3", "Python, VS Code", "Mantenimiento"],
              ]

def consultar_computadoras():
  print("\n---COMPUTADORAS---")
  for  i in range(len(computadoras)):
    print("Aula:", computadoras[i][0], "ID", computadoras[i][1],  "| software:", computadoras[i][2], "| Estado:", computadoras[i][3])



def buscar_sw():
  programa = input( "\nIngrese el software que desea buscar: ")
  print("\n--- RESULTADOS ---")
  encontrado = False
  for i in range(len(computadoras)):
      # la funcion (lower) convierte a minuscula el texto, de esta forma
      # nos aseguramos que el texto a buscar y la informacion de las listas
      # este en minuscula y no tengamos problema de que no encunentra un programa
      # porque este escrito de forma diferente usando mayusculas y minusculas
      if programa.lower() in computadoras[i][2].lower():
          print("ID:", computadoras[i][0], "| Aula:", computadoras[i][1],
                "| Software:", computadoras[i][2], "| Estado:", computadoras[i][3])

          encontrado = True

      if not encontrado == False:
        print("No se encontraron computadoras con ese software.")



def solicitar_prestamo():
  print("\n===SOLICITAR PRESTAMO===")
  num_control= input("Numero de control:")
  alumno_existe=False
  for alumno in alumnos:
    if alumno[i][0]==num_control:
      alumno_existe=True
      print(f"número de control: {alumno[0]}")
      print(f"Nombre del alumno: {alumno[1]}")
      print(f"Carrera: {alumno[2]}")
      break
  if alumno_existe:
    programa = input("Que programa necesita")
    for i in range(len(computadoras)):
      if programa.lower() in computadoras[i][2].lower():
        if computadoras[i][3]=="Disponible":
          computadoras[i][3]="Prestada"
          print("Se asignó la computadora", computadoras[i][1],
                "que esta en el aula", computadoras[i][0])
          prestamos.append([num_control, alumno[1], alumno[2], programa, computadoras[i][0],computadoras[i][1]])
          return
        else:
          print("Si se encontro el sofware pero esta", computadoras[i][3])
          return
    print("El software no existe en el sistema.")
    resp=input("¿Desea agregarlo a la lista de necesidades? (S/N): ")
    if resp.upper() == "S":
      fecha_hora = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
      necesidades.append([num_control, alumno[1], alumno[2], programa, fecha_hora])
      print("Se agregó el software a la lista de necesidades.")
    else:
      print("No se agregó el software a la lista de necesidades.")
  else:
    print("EL alumno no existe en la lista")


def devolver_computadora():
  print("\n ===DEVOLVER COMPUTADORA====")
  id_pc = input("Dame el ID de la computadora")
  existe = False
  for i in range( len(computadoras) ):
    if computadoras[i][0] == id_pc:
      existe = True
      if computadoras[i][3]== "Prestada" or computadoras[i][3]=="Reservada":
        computadoras[i][3] = "Disponible"
        print("La computadora", id_pc, "ya esta disponible")
        break
      else:
        print("La computadora", id_pc, "no esta prestada/reservada")
  if existe == False:
    print("No existe la computadora", id_pc)

def agregar_computadora():
  print("\n--- AGREGAR COMPUTADORA ---")
  id_computadora = input("ID de la computadora: ")
  aula = input("Aula: ")
  programas = input("Software instalado: ")
  nueva_pc=[id_computadora, aula, programas, "Disponible"]
  computadoras.append(nueva_pc)
  print("\nComputadora agregada correctamente.")

def historial_de_prestamos():
  print("===historial de prestamos===")
  for i in range(len(prestamos)):
    print(i+1, "| Alumnos", prestamos[i][0], "| sw",prestamos[i][1], "aula", prestamos[i][2],
          "ID", prestamos[i][3])

def consultar_necesidades():
  print("\n===LISTA DE NECESIDADES===")
  if len(necesidades) == 0:
      print("No hay necesidades registradas por el momento.")
  else:
      print("control\tNombre\tCarrera\tprograma\tFecha y Hora")
      print("-----------------------------------")
      for necesidad in necesidades:
        print(f"{necesidad[0]}\t{necesidad[1]}\t{necesidad[2]}\t{necesidad[3]}\t{necesidad[4]}")
#Menú usamos un ciclo while que se repetirá indefinidamente
opcion = -1
while opcion <= 8:
    print("\n================================")
    print("      SISTEMA DE PRÉSTAMOS")
    print("================================")
    print("1. Consultar computadoras")
    print("2. Buscar por software")
    print("3. Solicitar préstamo")
    print("4. Devolver computadora")
    print("5. Crear reservación")
    print("6. Agregar computadora")
    print("7. Historial de prestamos")
    print("8. Consultar necesidades")
    print("================================")
    opcion = int(input("Seleccione una opción: "))



    # 1. Consultar computadoras
    # si el usuario selecciono la opcion 1 se hace un ciclo (for) que recorre
    # toda la lista de computadoras e imprime los valores de las demás listas
    if opcion == 1:
      consultar_computadoras()

    # 2. Buscar software
    elif opcion == 2:
      buscar_sw()

    # 3. Solicitar prestamo
    elif opcion == 3:
      solicitar_prestamo()

    # 4. Devolver computadora
    elif opcion == 4:
      devolver_computadora()

    # 6. Agregar computadora
    elif opcion == 6:
      agregar_computadora()

    # 7. Historial de prestamos
    elif opcion == 7:
      historial_de_prestamos()

    # 8. Consultar necesidades
    elif opcion == 8:
      consultar_necesidades()
```
